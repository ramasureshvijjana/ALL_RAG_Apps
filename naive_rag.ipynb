```python
from langchain_community.document_loaders import TextLoader, WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
def load_docs(web_path:str = None, file_path:str = None):
    if web_path:
        web_loader = WebBaseLoader(web_path=(web_path))
        load_docs = web_loader.load()
    else:
        txt_loader = TextLoader(file_path)
        load_docs = txt_loader.load()
    return load_docs
```

```python
# Debuging the loaded docs
docs = load_docs(file_path = "E:\Gen_AI\ALL_RAG_Apps\Data\VitB3.txt")
print(docs)
print(docs[0])
# print(docs[0].page_content)
```
